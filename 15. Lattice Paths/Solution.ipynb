<p>How many such routes are there through a $20 \times 20$ grid?</p>

Starting in the top left corner of the grid you notice that you will always have to take 20 steps to the right and 20 step down, the only different is how you choose the ordering of those. This is a classical multiset problem where you have 40 items and 2 classes (down, right) with each class having 20 elements. Order within those classes is not important, but order between the classes is important. So we elemenate the order within classes by dividing by 20! for each class.

```python
def factorial(n):
    if n < 0:
        return 0

    res = 1
    for i in range(n):
        res = (i + 1) * res

    return res
```

```python
answer = factorial(40)//(factorial(20)**2)
print(answer)
```
